### Importer les données

## Question 1  

```python
df_ks.sort_values(by='usd_pledged_real', ascending=False).head(5)
df_ks[['name', 'usd_pledged_real']].head(5)
```

## Question 2

```python
# Convert to float for comparison
df_ks['usd_pledged_real'] = pd.to_numeric(df_ks['usd_pledged_real'], errors='coerce')
df_ks['goal'] = pd.to_numeric(df_ks['goal'], errors='coerce')

# Count projects that reached their goal
projects_reached_goal = (df_ks['usd_pledged_real'] >= df_ks['goal']).sum()
print(f"Nombre de projets ayant atteint leur but: {projects_reached_goal}")
```

## Question 3

```python
# Count projects by category
projects_by_category = df_ks['category'].value_counts().sort_values(ascending=False)
print("Nombre de projets par catégorie:")
print(projects_by_category)
```

## Question 4

```python
# Convert launched column to datetime
df_ks['launched'] = pd.to_datetime(df_ks['launched'], errors='coerce')

# Filter French projects created before 2016
french_projects_before_2016 = df_ks[(df_ks['country'] == 'FR') & (df_ks['launched'].dt.year < 2016)]
count_french_before_2016 = len(french_projects_before_2016)
print(f"Nombre de projets français instanciés avant 2016: {count_french_before_2016}")
print(f"Projets: {french_projects_before_2016[['name', 'country', 'launched']].head()}")
```

## Question 5

```python
# Get American projects that asked for more than 200,000 dollars
american_projects_200k = df_ks[(df_ks['country'] == 'US') & (df_ks['goal'] > 200000)]
print(f"Nombre de projets américains ayant demandé plus de 200 000 dollars: {len(american_projects_200k)}")
print("\nProjets américains avec goal > $200,000:")
print(american_projects_200k[['name', 'country', 'goal', 'usd_pledged_real']])
```

## Question 6 

```python
# Count projects with "Sport" in their name
sport_projects = df_ks[df_ks['name'].str.contains('Sport', case=False, na=False)]
count_sport = len(sport_projects)
print(f"Nombre de projets ayant 'Sport' dans leur nom: {count_sport}")
print("\nProjets avec 'Sport' dans le nom:")
print(sport_projects[['name', 'category', 'goal']])
```
